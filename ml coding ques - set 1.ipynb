```python
#1#Question 1:
#A retail store wants to identify customers who make frequent purchases. Given the dataset below, write a Python program to:
#Group customers by their IDs.
import warnings
warnings.filterwarnings('ignore')

import pandas as pd

# 1. Define the dataset
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

# 2. Create the DataFrame
df = pd.DataFrame(data)

# 3. Group by Customer_ID and calculate total purchases
total_purchases = df.groupby('Customer_ID')['Purchase_Amount'].sum().reset_index()

# 4. Find the top 3 frequent customers
top_customers = total_purchases.sort_values(by='Purchase_Amount', ascending=False).head(3)

# 5. Display the results
print("Total Purchases per Customer:")
print(total_purchases)
print("\nTop 3 Frequent Customers:")
print(top_customers)
```

```python
#Third question below scenario

#Identifying Frequent Labels in a Dataset
#A company wants to identify the top 3 most common categories in a dataset. Given the dataset below, write a Python program to:
#Group the data by Category.

import pandas as pd

# 1. Define the dataset
data = {
    'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
    'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]
}

# 2. Create the DataFrame
df = pd.DataFrame(data)

# 3. Group by Category and count occurrences
category_counts = df.groupby('Category').size().reset_index(name='Count')

# 4. Identify the top 3 most common categories
top_3_categories = category_counts.nlargest(3, 'Count')

# 5. Display the results
print("Category Counts:")
print(category_counts)
print("\nTop 3 Most Common Categories:")
print(top_3_categories)
```

```python
#4.Predicting Missing Values Using Mean Imputation A dataset contains missing values in the Age column. Write a Python program to:
#Replace missing values with the mean of the column.


import pandas as pd

# A dataset contains missing values in the Age column. Write a Python program to:
# Replace missing values with the mean of the column.

# 1. Define the dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, None, 35, None]
}

# 2. Create the DataFrame
df = pd.DataFrame(data)

# 3. Calculate the mean age (ignoring the missing values)
mean_age = df['Age'].mean()

# 4. Fill the missing values with the calculated mean
df['Age'] = df['Age'].fillna(mean_age)

# 5. Display the result
print(df)


```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Implementing a Simple Linear Regression Model
# Dataset with Experience (years) and Salary ($)
# Task: Train a Linear Regression model.

# 1. Define the dataset
data = {'Experience': [1, 2, 3, 4, 5], 
        'Salary': [30000, 35000, 40000, 45000, 50000]}

# 2. Convert to DataFrame and separate feature (X) and target (y)
df = pd.DataFrame(data)
X = df[['Experience']]  # Features must be a 2D structure (DataFrame)
y = df['Salary']        # Target variable

# 3. Train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

print("Linear Regression model trained successfully!")
```
