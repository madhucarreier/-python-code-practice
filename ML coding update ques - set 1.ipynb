```python
#1.MAdhu Answer
#Question 1:A retail store wants to identify customers who make frequent purchases. Given the dataset below, write a Python program to:
#Group customers by their IDs.


import warnings
warnings.filterwarnings('ignore')

import pandas as pd

# 1. Define the dataset
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

# 2. Create the DataFrame
df = pd.DataFrame(data)

# 3. Group by Customer_ID and calculate total purchases
total_purchases = df.groupby('Customer_ID')['Purchase_Amount'].sum().reset_index()

# 4. Find the top 3 frequent customers
top_customers = total_purchases.sort_values(by='Purchase_Amount', ascending=False).head(3)

# 5. Display the results
print("Total Purchases per Customer:")
print(total_purchases)
print("\nTop 3 Frequent Customers:")
print(top_customers)
```

```python
#1(3)Identify the top 3 customers with the highest purchase amounts.
import pandas as pd

# 1. Define the dataset
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103], 
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}

# 2. Create the DataFrame
df = pd.DataFrame(data)

# 3. Calculate total purchase amount per customer
total_spent = df.groupby('Customer_ID')['Purchase_Amount'].sum().reset_index()

# 4. Identify the top 3 customers with the highest purchase amounts
top_3_customers = total_spent.nlargest(3, 'Purchase_Amount')

# 5. Display the result
print("Top 3 Customers with Highest Purchase Amounts:")
print(top_3_customers)
```

```python
#2(1)A real estate company wants to predict house prices based on square footage. Write a Python program to:
#Train a Linear Regression model.

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Define the dataset
data = {
    'Square_Feet': [1500, 2000, 2500, 3000, 3500],
    'Price': [300000, 400000, 500000, 600000, 700000]
}

# 2. Convert to a pandas DataFrame
df = pd.DataFrame(data)

# 3. Separate features (X) and target variable (y)
# Scikit-learn expects X to be a 2D array or DataFrame
X = df[['Square_Feet']]
y = df['Price']

# 4. Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# 5. Display the learned parameters
print(f"Model Training Complete!")
print(f"Intercept (Base Price): ${model.intercept_:.2f}")
print(f"Coefficient (Price per Sq. Ft.): ${model.coef_[0]:.2f}")

# Example: Predict the price of a 2,800 sq. ft. house
sample_sq_ft = [[2800]]
predicted_price = model.predict(sample_sq_ft)
print(f"Predicted price for 2,800 sq. ft.: ${predicted_price[0]:.2f}")
```

```python
#3(1):Identifying Frequent Labels in a Dataset
#A company wants to identify the top 3 most common categories in a dataset. Given the dataset below, write a Python program to:
#Group the data by Category.


import pandas as pd

# 1. Define the dataset
data = {
    'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
    'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]
}

# 2. Convert to DataFrame
df = pd.DataFrame(data)

# 3. Group by Category and count frequencies
# Sorting in descending order places the most common categories at the top
category_counts = df.groupby('Category').size().sort_values(ascending=False)

# 4. Extract the top 3 categories
top_3_categories = category_counts.head(3)

print("Top 3 Most Common Categories:")
print(top_3_categories)



```

```python
#4(1)Predicting Missing Values Using Mean Imputation
#A dataset contains missing values in the Age column. Write a Python program to:
#Replace missing values with the mean of the column.

import numpy as np
import pandas as pd

# 1. Define the original dataset with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, None, 35, None]
}

# 2. Create the DataFrame (None becomes NaN automatically)
df = pd.DataFrame(data)

print("--- Original Data ---")
print(df)

# 3. Calculate the mean of the Age column (ignores NaN values)
age_mean = df['Age'].mean()

# 4. Fill the NaN values with the calculated mean
df['Age'] = df['Age'].fillna(age_mean)

print(f"\nCalculated Mean Age: {age_mean}")
print("\n--- Imputed Data ---")
print(df)
```

```python
#5(2) Predict the salary for an individual with 6 years of experience
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Define the dataset
data = {
    'Experience':[1, 2, 3, 4, 5], 
    'Salary': [30000, 35000, 40000, 45000, 50000]
}
df = pd.DataFrame(data)

# 2. Split into feature (X) and target (y)
X = df[['Experience']]
y = df['Salary']

# 3. Train the model
model = LinearRegression()
model.fit(X, y)

# 4. Predict for 6 years of experience
predicted_salary = model.predict([[6]])

# 5. Display the result
print(f"Predicted Salary for 6 years of experience: ${predicted_salary[0]:.0f}")

```
